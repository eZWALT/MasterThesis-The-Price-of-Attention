## 1.2 Recommender Systems Taxonomy

### Types of Recommendations

Depending on the type of recommendations the data and modelling varies drastically, to have a big picture in mind and have a good taxonomy:

**Firstly**, depending on the type of filtering (Roots of the tree):
1. Collaborative Filtering (CF): can be user-focused (find similar users) or item-focused (find similar or bought-together products) and works by association. 
2. Content-Based Filtering (CB): relies on metadata (tags, description, category)
3. Hybrid Systems: Specially useful to solve cold-start; weighted averages of CB/CF, ...

**Secondly**, depending on the dimensional structure of the data (time / space) (Trunk of the Tree):
1. Session-based Recommendation (Sequential-burst)
2. History-based Recommendation (Life Story)
    - 2.1 Static (User-Item matrices)
    - 2.2 Dynamic (Sequences)
3. Hybrid: Session-Aware Recommendation
4. Graph-Based: Modelling users and items in graphs.

**Thirdly**, more specialized branches of recommendation systems (Branches of the Tree):
1. Context Aware RS (CARS)
2. Knowledge Aware RS (KARS)
3. Conversational
4. Multi-Criteria (MCRS)
5. Cross-Domain: Using spotify data to recommend on Amazon...

In terms of using RAG, the representation layer becomes the bottleneck for session-based recommendation and generation becomes the bottleneck of history-based recommendation. The sequential recommendations have been more popular in the literature, but a growing trend of session-based RAG RecSys has appeared lately.

> **Possible Open Problem**: Session-Based RAG Recommendations

### RAG Paths for RecSys

- **The Representation Path**: Using RAG to build "Smarter States" for RL (solving the cold start).
- **The Generation Path**: Using LLMs to explain why an RL agent chose a specific (and perhaps weird) item.

## 1.4 Challenges

The main challenges that **traditional** recommender systems face:

| Challenge | Affected Systems |
|---|---|
| Cold-start | CF, CB, Session-Based |
| Filter Bubble | CB |
| Vertical Scalability | CF |
| Popularity Bias | CF |
| Lack of Explainability | CF |

**Modern** RecSys integrating deep learning, LLMs, MLLMs and agentic approaches face additional challenges:

| Challenge | Description |
|---|---|
| Long-term vs Short-term | Exploitation–Exploration RL dilemma |
| Multi-Objective trade-off | Balancing revenue, trust, diversity, etc. |
| Ethical concerns & fairness | Demographic biases |
| Sparsity of feedback | Users interact with only a small portion of items |

## 1.5 Reinforcement Learning in RecSys

Reinforcement learning approaches are good at leveraging the exploitation vs exploration dilemma. Regular ML is greedy and exploitative, thus RL can help avoid the filter bubble.

### Outlook (LLM Opinion... sus)

By 2026, the intersection of LLMs and Recommender Systems has matured into a complex, agent-driven ecosystem that prioritizes human-aligned reasoning and industrial efficiency. The shift from user-based to item-based RAG, the standardization of CALM-style hybrid architectures, and the emergence of neurophysiological evaluation metrics signify a move toward more robust and transparent systems. However, several challenges remain. The potential for LLM-induced "psychogenicity"—where models might validate a user's harmful or delusional beliefs in a conversational setting—requires rigorous safety benchmarking. Furthermore, the risk of "circularity" in evaluation, where LLMs grade other LLMs, necessitates continued human-in-the-loop validation and the development of truly independent metrics.

The future of CRS lies in "Vibe-Coding" for rapid system iteration and the further convergence of voice input with emotional markup, allowing agents to detect and respond to subtle cues in a user's tone or speech pattern. As hardware capabilities and model efficiency continue to improve, the integration of real-time multi-step reasoning into everyday consumer interfaces will become the standard, finally fulfilling the promise of truly intelligent, conversational, and personalized digital assistants. The ongoing success of these systems will depend on their ability to balance the powerful generative capabilities of LLMs with the deterministic constraints and ethical requirements of the domains they serve.

    > BLIND SPOTS: Cold-start (No collaborations yet), Evidence graphs doesn't scale well, RecSys and RL won't go hand in hand because of sparsity, popularity based warm start is used but future work could be using synthetic rewards :)

## 2.3 RAG for RecSys — 2025

- **ARAG (25)**: Agentic-RAG to build a conversational - session-based RecSys using 4 LLM's agents to perform both history-based and session-based hybrid recommendations, using:
    * User summarizer
    * Semantic Inference
    * Ranking
    * ...
    
    To avoid query ambiguity were traditional RAG fails miserably. Users often give vague intents (e.g., "I want a new hobby"). A standard RAG just searches "new hobby." for example.

    BLIND SPOTS: Ultra slow due to agentic workflow, possible hallucation and agentic trajectory could be optimized (10 steps down to 5...)

- **SarRec (25)**: retrieval schema (WHICH SHOE I SHOULD SHOW TO THE USER) that improves RAG recommendations (SASRec for example) were recommendations are not only done via IDs (user A clicked item 1) but adding semantic dia via RAG for SASRec  to improve representation. 

    This is the founding father of RAG - RecSys
    BLIND SPOTS: Rag Noise, lacks end to end optimization and lack of multi-objective.



- **CatalogRag (25)**: (WHAT TAGS SHOULD BE ASSOCIATED TO THIS SHOE????) Amazon's multilanguage RAG to perform CB recommendation that makes use of their catalogs to improve not only RAG performance but LLM prediction of products attributes.


    BLIND SPOTS: Hallucination can spread over the database, brand new categories probably has cold start (not good for novel items).

- **Federated Recs with Hybrid RAG - GPT-FedRec (25)**: uses federated learning to solve privacy issues and Hybrid RecSys (CB semantically + CF ids sequence) to retrieval and then zero-shot reranking via LLM. 

    BLIND SPOTS: Weight of CB/CF, communication cost (possible compression of descriptions?), local bias due to RAG (things that exist vs things that doesnt)

- **Collaborative Retrieval for conversational recsys (CRAG) (25)**: Performs context aware retrieval for behavioral physics (90% of people who likes nike also likes MyProtein) that will be feed into a conversational multi-turn LLM, so Behaviour blind spot of LLMs is covered, because they are good in dialogue already.

- **KG-RAG for LLM-based recs KGRagRec (25)**: For history-based recommendations, normal RAG's can be noisy, outdated and hallucinated, by using knowledge graphs we can get faithful and related knowledge to improve world representation.

- **LlamaREC-LKG-RAG (25)**: KG help go beyond similar items (filter bubble) but dumping the whole KG into a RAG generates noise, and also LLM-reranking is slow. So a sequential session-based recsys is created with a learnable KG tailored to the user. **Way faster because its logit based instead of text based**

## 2.4 Advertising in LLMs

> Advertising originates from the Latin term advertere, meaning "to turn (attention) toward" or "to make aware". 

- **Online Advertisement With LLMs (2023)**: Proposes the 1st framework **(LAS)** to bridge genAI and ads, shifting from keyword matching to output modification. Also tries to tackle the new difficulties that appear in bidding from keywords to generation probabilities. The 2 main contributions are the introduction of Dynamic Creative Optimization (DCO, responsive ad) and the LAS framework that uses 4 modules to achieve this:

    * **Modification**: Multiple generations with potential ads
    * **Bidding**: Advertisers bid on specific modified outputs
    * **Prediction**: Estimate user satisfaction and CTR for each modified output (Balance...)
    * **Auction**: Maximizes the original formula for advertisement presented below and determines which add is displayed and how much it pays.

    \begin{aligned}
    \max_{x_{\text{ad},\text{slot}}} \quad 
    & \sum_{\text{ad} \in \mathcal{Ads}} \sum_{\text{slot} \in \mathcal{Slots}} 
    x_{\text{ad},\text{slot}} \cdot \text{bid}_{\text{ad}} \cdot \mathrm{pCTR}_{\text{ad},\text{slot}} \\[6pt]
    \text{s.t.} \quad 
    & \sum_{\text{ad} \in \mathcal{Ads}} x_{\text{ad},\text{slot}} \le 1 
    \quad \forall \, \text{slot} \in \mathcal{Slots} \\[4pt]
    & \sum_{\text{slot} \in \mathcal{Slots}} x_{\text{ad},\text{slot}} \le 1 
    \quad \forall \, \text{ad} \in \mathcal{Ads} \\[4pt]
    & x_{\text{ad},\text{slot}} \in \{0,1\}
    \quad \forall \, \text{ad} \in \mathcal{Ads}, \forall \, \text{slot} \in \mathcal{Slots}
    \end{aligned}

   This paper is an amazing baseline (conceptually speaking... no implementation) and gives rise to multiple questions or details we could further explore in our own research:
   * CTR & SR (satisfaction rate) are proxy metrics to true objectives, which are those objectives and is there a better to proxy metric to approximate it?
   * There are 3 kind of ads to take into account: Regular ads, Conversational ads (dyanmically tailored DCO or not) and the current approach that OpenAI takes that is 
   * This 4 module pipeline is too rigid, by relaxing the constraints we can improve latency by removing bottlenecks such as the modification module, we could use prediction models to speed up things but removing real-time control from advertisers.
   * How can we estimate satisfaction rate? One way is to assume original output X is optimal, then $d(X, X_i)$ and online-learning model can suffice.

   BLIND SPOTS: Latency when generating (multiple responses + real-time auction), privacy risks for user, difficult interpolation (ads are not being mixed) and hallucation will erode user trust...

**The death of the click & new metrics:**

- Clicks are disappearing — users are lazy, they stay in-chat.
- New pricing paradigms are emerging:
  - **Cost per Turn (CPT)** — how many turns after an ad is shown does the user keep engaging?
  - **Attention Shift** — if a fitness conversation leads to a protein recommendation, *how much does the user's attention shift toward protein?* This is the purest proxy for ad quality.

**GEO — Generative Engine Optimization:**

- Overpromising will backfire: users now have authoritative LLM-backed recommendations.
- Advertisers must optimise product descriptions for **LLM understanding**: accurate titles, clean attributes, strong imagery, correct pricing.
- **Transparency & quality** will always win; both sides lose trust otherwise.
- Fewer, more accurate ads > 1 000 spammy ones.

# 5. Mathematical Formulation

The original formula that is maximized in regular advertisement doesnt contain multiple objectives, this could be useful for multi-objective optimization scenarios to be thought:

\begin{aligned}
\max_{x_{\text{ad},\text{slot}}} \quad 
& \sum_{\text{ad} \in \mathcal{Ads}} \sum_{\text{slot} \in \mathcal{Slots}} 
x_{\text{ad},\text{slot}} \cdot \Big(
\alpha \cdot \text{bid}_{\text{ad}} \cdot \mathrm{pCTR}_{\text{ad},\text{slot}} 
+ \beta \cdot \mathrm{pConversion}_{\text{ad},\text{slot}} 
+ \gamma \cdot \mathrm{UserValue}_{\text{ad},\text{slot}}
+ ...
\Big) \\[6pt]
\text{s.t.} \quad 
& \sum_{\text{ad} \in \mathcal{Ads}} x_{\text{ad},\text{slot}} \le 1 
\quad \forall \, \text{slot} \in \mathcal{Slots} \\[4pt]
& \sum_{\text{slot} \in \mathcal{Slots}} x_{\text{ad},\text{slot}} \le 1 
\quad \forall \, \text{ad} \in \mathcal{Ads} \\[4pt]
& x_{\text{ad},\text{slot}} \in \{0,1\}
\quad \forall \, \text{ad} \in \mathcal{Ads}, \forall \, \text{slot} \in \mathcal{Slots}
\end{aligned}
